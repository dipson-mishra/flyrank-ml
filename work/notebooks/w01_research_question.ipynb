## 3. Quick look at the data (2-3 real numbers)

The starter dataset shows a significant opportunity for refresh work. Over half of the pages are currently in a state of decline, and those declining pages still maintain substantial visibility, making them prime candidates for recovery efforts.

```python
import pandas as pd

# Load the starter CSV
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# Analyze declining pages
declining = df[df['trend_direction'] == 'down']
total = len(df)
num_declining = len(declining)
pct_declining = (num_declining / total) * 100
avg_imp_declining = declining['impressions_90d'].mean()
avg_imp_not_declining = df[df['trend_direction'] != 'down']['impressions_90d'].mean()

print(f"Total pages in dataset: {total}")
print(f"Pages with 'down' trend: {num_declining} ({pct_declining:.2f}%)")
print(f"Avg impressions (declining): {avg_imp_declining:.2f}")
print(f"Avg impressions (not declining): {avg_imp_not_declining:.2f}")
```
