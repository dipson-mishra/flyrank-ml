## 3. Quick look at the data (2-3 real numbers)

Looking at the starter dataset, there is a massive volume of pages with significant exposure (impressions) but very low engagement. A large portion of the visible inventory has a CTR well below 0.5%, suggesting a systemic opportunity to improve the "capture rate" of existing search visibility.

```python
import pandas as pd

# Load the starter CSV
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# Define 'visible' as pages with at least 500 impressions in 90 days
visible = df[df['impressions_90d'] >= 500]
# Identify 'under-capturing' pages as those with CTR < 0.5%
low_ctr = visible[visible['ctr'] < 0.5]

print(f"Total visible pages (imp >= 500): {len(visible)}")
print(f"Pages with low CTR (<0.5%): {len(low_ctr)}")
print(f"Avg CTR of all visible pages: {visible['ctr'].mean():.2f}%")
print(f"Avg CTR of low-CTR visible pages: {low_ctr['ctr'].mean():.2f}%")
```
